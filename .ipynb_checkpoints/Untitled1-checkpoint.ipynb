```python
import sys
sys.path.insert(0, '/home/yuandu/cpp_base/build')
import cpp_core

print("=== ThreadSafeStack 测试 ===\n")

# 创建栈
s = cpp_core.ThreadSafeStack()
print("1. 创建空栈:", s)
print("   是否为空?", s.empty())

# 入栈
s.push(10)
s.push(20)
s.push(30)
print("\n2. 入栈 10, 20, 30 后:", s)
print("   是否为空?", s.empty())

# 出栈
val = s.pop()
print("\n3. 出栈:", val)
print("   出栈后栈:", s)

# 再出栈
val = s.pop()
print("\n4. 再出栈:", val)
print("   再出栈后栈:", s)

# 测试空栈出栈
val = s.pop()
print("\n5. 第三次出栈:", val)
print("   第三次出栈后栈:", s)

val = s.pop()
print("\n6. 空栈出栈:", val)
print("   空栈出栈后栈:", s)
```
